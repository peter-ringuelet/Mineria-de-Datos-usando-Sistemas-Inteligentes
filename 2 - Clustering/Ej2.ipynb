```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x1 = np.array([2, 0])
x2 = np.array([1, 2])
x3 = np.array([3, 4])
x4 = np.array([4, 4])

df = pd.DataFrame({'x': [x1[0], x2[0], x3[0], x4[0]], 'y': [x1[1], x2[1], x3[1], x4[1]]})

# Aplicar K-Means con K=2
kmeans = KMeans(n_clusters=2, random_state=42, n_init=1, verbose=1, max_iter=3)
df['Cluster'] = kmeans.fit_predict(df)

# Obtener los centroides
centroids = kmeans.cluster_centers_
print("Centroides:")
print(pd.DataFrame(centroids, columns=['X1', 'X2']))

# Diagrama de dispersión
plt.figure(figsize=(6, 4))
sns.scatterplot(x=df['x'], y=df['y'], hue=df['Cluster'].astype(str), palette='Set1')
plt.scatter(centroids[:, 0], centroids[:, 1], c='black', marker='X', s=200, label='Centroides')
plt.legend(title="Clase")
plt.title("Diagrama de dispersión de los puntos")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
