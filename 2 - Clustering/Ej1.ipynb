```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## a) Calcular la distancia euclídea entre los puntos p1 = (2,-2,2,-4) y p2 = (3,0,6,-2). Calcular también la distancia euclídea al cuadrado.

```python
# Definimos los puntos
p1 = np.array([2, -2, 2, -4])
p2 = np.array([3, 0, 6, -2])

# Calculamos la distancia euclídea
distancia_euclidea_p1_p2 = np.sqrt(np.sum((p1 - p2)**2))
print(f"Distancia euclídea entre p1 y p2: {distancia_euclidea_p1_p2}")

# Calculamos la distancia euclídea al cuadrado
distancia_euclidea_cuadrado_p1_p2 = np.sum((p1 - p2)**2)
print(f"Distancia euclídea al cuadrado entre p1 y p2: {distancia_euclidea_cuadrado_p1_p2}")
```

### b) Dados los centroides c1 y c2 de la Figura 2, calcular las distancias euclídea del punto x1 a c1 y c2. Luego indicar el centroide más cercano a x1. Calcular también la distancia euclídea al cuadrado. 
- c1 = (0,3)  
- c2 = (-1, 1)
- x1 = (2, 1)

```python
# Definimos los puntos
c1 = np.array([0, 3])
c2 = np.array([-1, 1])
x1 = np.array([2, 1])

# Calculamos la distancia euclídea entre x1 y c1
distancia_euclidea_x1_c1 = np.sqrt(np.sum((x1 - c1)**2))
print(f"Distancia euclídea entre x1 y c1: {distancia_euclidea_x1_c1}")

# Calculamos la distancia euclídea entre x1 y c2
distancia_euclidea_x1_c2 = np.sqrt(np.sum((x1 - c2)**2))
print(f"Distancia euclídea entre x1 y c2: {distancia_euclidea_x1_c2}")

# Determinamos el centroide más cercano a x1
if distancia_euclidea_x1_c1 < distancia_euclidea_x1_c2:
    print("El centroide más cercano a x1 es c1")
else:
    print("El centroide más cercano a x1 es c2")

# Calculamos la distancia euclídea al cuadrado entre x1 y c1
distancia_euclidea_cuadrado_x1_c1 = np.sum((x1 - c1)**2)
print(f"Distancia euclídea al cuadrado entre x1 y c1: {distancia_euclidea_cuadrado_x1_c1}")

# Calculamos la distancia euclídea al cuadrado entre x1 y c2
distancia_euclidea_cuadrado_x1_c2 = np.sum((x1 - c2)**2)
print(f"Distancia euclídea al cuadrado entre x1 y c2: {distancia_euclidea_cuadrado_x1_c2}")
```

### c) La distancia de Manhattan entre dos puntos A y B se define como d(A,B) = Σi  |Ai-Bi|, donde Ai y Bi son los valores de  los puntos. Por ejemplo, si A=(3,0,1) y B=(-2,4,7), entonces d(A,B)= |3-(-2)| + |0-4| + |1-7| = |5| + |-4| + |-6| = 5 + 4 + 6 = 15
Repetir los puntos a-b con la distancia de Manhattan.

```python
distancia_manhattan_p1_p2 = np.sum(np.abs(p1-p2))
print(f"Distancia de Manhattan entre p1 y p2: {distancia_manhattan_p1_p2}")

distancia_manhattan_x1_c1 = np.sum(np.abs(x1-c1))
print(f"Distancia de Manhattan entre x1 y c1: {distancia_manhattan_x1_c1}")

distancia_manhattan_x1_c2 = np.sum(np.abs(x1-c2))
print(f"Distancia de Manhattan entre x1 y c2: {distancia_manhattan_x1_c2}")

```

```python
tabla_comparacion = pd.DataFrame({
    'Puntos': ['p1,p2', 'x1,c1', 'x1,c2'],
    'Distancia Euclídea': [distancia_euclidea_p1_p2, distancia_euclidea_x1_c1, distancia_euclidea_x1_c2],
    'Distancia Euclídea al cuadrado': [distancia_euclidea_cuadrado_p1_p2, distancia_euclidea_cuadrado_x1_c1, distancia_euclidea_cuadrado_x1_c2],
    'Distancia de Manhattan': [distancia_manhattan_p1_p2, distancia_manhattan_x1_c1, distancia_manhattan_x1_c2]
})
display(tabla_comparacion)
```

### d) ¿Qué relación encuentra entre la distancia Euclídea y la Euclídea al cuadrado, para las distancias entre x1 y c1 y  entre x1 y c2?
Que se mantiene la relacion de orden

```python
x1 = np.array([2, 3])
x2 = np.array([-2, 2])
x3 = np.array([6, -14])
#  calcular el centroide C de los mismos. Dibujar los datos y el centroide C en un plano cartesiano.

# Calcular el centroide C de los puntos
centroide = np.mean(np.array([x1, x2, x3]), axis=0)
print(f"Centroide C: {centroide}")

# Crear el plano cartesiano
fig, ax = plt.subplots()

# Dibujar los puntos
ax.scatter(x1[0], x1[1], color='red', label='x1')
ax.scatter(x2[0], x2[1], color='blue', label='x2')
ax.scatter(x3[0], x3[1], color='green', label='x3')

# Dibujar el centroide C
ax.scatter(centroide[0], centroide[1], color='black', label='Centroide C')

# Mostrar la gráfica
plt.legend()
plt.show()

```
